# KNN - K-NEAREST NEIGHBOUR

```python
from sklearn.datasets import load_iris
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score

from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target, name='species')
# setosa
# versicolor
# virginica

# sepal length (cm)
# sepal width (cm)
# petal length (cm)
# petal width (cm)
```

```python
iris
```

```python
print("Target classes:", iris.target_names)
```

```python
target_mapping = {i: str(name) for i, name in enumerate(iris.target_names)}
print(target_mapping)
```

```python
X_df = X.copy()
X_df['Actual'] = y
```

```python
print("Dataset shape:", X_df.shape)
```

```python
X_df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2, stratify=y
)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
k = 5
knn = KNeighborsClassifier(n_neighbors=k)
```

```python
knn.fit(X_train_scaled, y_train)
```

```python
y_train_pred = knn.predict(X_train_scaled)
y_test_pred = knn.predict(X_test_scaled)
```

```python
y_train_pred
```

```python
y_test_pred
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Train Accuracy:", accuracy_score(y_train, y_train_pred))
print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
```

```python
def predict_iris(sepal_length, sepal_width, petal_length, petal_width):
    sample_df = pd.DataFrame([[sepal_length, sepal_width, petal_length, petal_width]], columns=['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)'])
    sample_scaled = scaler.transform(sample_df)
    prediction = knn.predict(sample_scaled)[0]
    predicted_species = iris.target_names[prediction]
    return predicted_species, prediction
```

```python
pred, prediction = predict_iris(5.1, 3.5, 1.4, 0.2)
print("\nExample Prediction -> Species:", pred)
```

```python
predictions = []
predictions_indx = []

for i in X.values:
    pred, predictions_index = predict_iris(i[0], i[1], i[2], i[3])
    predictions.append(pred)
    predictions_indx.append(predictions_index)
```

# Advanced Cross Validation : To find best K-value

```python
X_df.to_csv("KNN_k_5_irirs_predicted_value.csv")
```

```python
import time
```

```python
cv_scores = []
K_range = range(1,21)
for k in K_range:
    knn_cv = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn_cv, X, y, cv=10, scoring = 'accuracy')
    print(scores)
    time.sleep(1)
    cv_scores.append(scores.mean())
```

```python
import numpy as np
```

```python
best_k = K_range[np.argmax(cv_scores)]
```

```python
print(f"Best K using Cross Validation : {best_k}")
print(f"Cross Validation accuracy for K={best_k}: {max(cv_scores):.4f}")
```
