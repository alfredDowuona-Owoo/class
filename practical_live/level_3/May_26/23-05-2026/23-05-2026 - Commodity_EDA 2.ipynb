# Commodity Prices Dataset Analysis

Kaggle Dataset link : https://www.kaggle.com/datasets/debashish311601/commodity-prices

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("data/commodity_futures.csv")
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values("Date").reset_index(drop=True)
df.head()
```

```python
# Create Year & Month for seasonal/group analysis
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df.head()
```

```python
# Quick check of missing values
print(df.isna().sum())
```

```python
# Forward fill missing values
df_ffill = df.ffill()
```

```python
# Quick check of missing values
print(df_ffill.isna().sum())
```

```python
df_clean = df.ffill().bfill()
```

```python
# Check again
print(df_clean.isna().sum())
```

### Descriptive stats

```python
# Price Trends of Commodities (Descriptive)
## Metals
metals = ['ALUMINIUM', 'NICKEL', 'ZINC', 'COPPER', 'GOLD', 'SILVER']
plt.figure(figsize=(14,6))
for col in metals:
    plt.plot(df_clean['Date'], df_clean[col], label=col)
plt.title("Commodity Price Trends (2000–2024)")
plt.xlabel("Year")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
# Price Trends of Commodities (Descriptive)
## Metals without Nickel
metals = ['ALUMINIUM',  'ZINC', 'COPPER', 'GOLD', 'SILVER']
plt.figure(figsize=(14,6))
for col in metals:
    plt.plot(df_clean['Date'], df_clean[col], label=col)
plt.title("Commodity Price Trends (2000–2024) (without Nickel)")
plt.xlabel("Year")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
# Price Trends of Commodities (Descriptive)
## Metals just copper and silver
metals = ['COPPER', 'SILVER']
plt.figure(figsize=(14,6))
for col in metals:
    plt.plot(df_clean['Date'], df_clean[col], label=col)
plt.title("Commodity Price Trends (2000–2024) (Only Copper and Silver)")
plt.xlabel("Year")
plt.ylabel("Price")
plt.legend()
plt.show()
```

## Seasonal & Volatility Analysis

```python
import matplotlib.pyplot as plt
```

### Seasonality

```python
yearly_avg = df_clean.groupby("Year").mean()
```

```python
yearly_avg
```

```python
# Yearly Average Prices
# See long-term upward/downward movement

plt.figure(figsize=(12,6))
for col in ["GOLD", "WTI CRUDE", "SOYBEANS", "COPPER"]:
    plt.plot(yearly_avg.index, yearly_avg[col], label=col)
plt.title("Yearly Average Commodity Prices")
plt.xlabel("Year")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
# Monthly Seasonality
# Are some months consistently higher/lower?

monthly_avg = df_clean.groupby("Month").mean()

plt.figure(figsize=(12,6))
for col in ["GOLD", "WTI CRUDE", "SOYBEANS", "COPPER", "COTTON"]:
    plt.plot(monthly_avg.index, monthly_avg[col], label=col)
plt.title("Monthly Average Commodity Prices (Seasonality)")
plt.xlabel("Month")
plt.ylabel("Price")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
for col in ["SUGAR", "CORN", "SOYBEANS", "COPPER", "COTTON", "COFFEE", "SOYBEAN OIL", "HRW WHEAT"]:
    plt.plot(monthly_avg.index, monthly_avg[col], label=col)
plt.title("Monthly Average Commodity Prices (Seasonality)")
plt.xlabel("Month")
plt.ylabel("Price")
plt.legend()
plt.show()
```

### Volatility
- Standard deviation of returns to compare how risky or volatile prices changes.
 - it looks at the deviation of prices from the mean

```python
# Volatility
## pc_change() looks at percentage change or how much spread each commodity has
returns = df_clean.set_index("Date").pct_change().dropna()
```

```python
returns.head()
```

```python
print(returns.std())
```

```python
volatility = returns.std().sort_values(ascending=False)
```

```python
volatility
```

```python
plt.figure(figsize=(10,6))
volatility_new = volatility.drop(['Month', 'Year'])
volatility_new.head(10).plot(kind="bar", color="salmon")
plt.title("Top 10 Most Volatile Commodities")
plt.ylabel("Std Dev of Daily Returns")
plt.show()
```
